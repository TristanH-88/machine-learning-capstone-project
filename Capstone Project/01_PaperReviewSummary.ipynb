# Milestone 1: (Research Paper Review) - Mining Big Data to Extract Patterns

The authors of the research paper introduce a variety of analytical methods
 and algorithms designed to help social scientists extract insights
 from large-scale digital footprints. The authors utilize a dataset of digital footprints and psychodemographic profiles to demonstrate their analytical methods. They sample data from U.S. Facebook users and their associations with Facebook likes. They use three csv files including users.csv, likes.csv, and users-likes.csv.

The users.csv file include Facebook profiles of 110,728 users.
 It includes nine columns: an anonymized user ID, gender, age, political views,
 and five personality scores based on the Five-Factor Model
  (Openness, Conscientiousness, Extroversion, Agreeableness, and Neuroticism).
  The likes.csv file includes lists of 1,580,284 unique Facebook Likes.
  It contains two columns: an anonymized Like ID and the name of the Like.
  The users-likes file records the associations between users and their Likes.
  It contains over 10.6 million rows and two columns: user ID and Like ID.

The authors used Regression Models, Cross-Validation, Singular Value Decomposition,
User–Footprint Matrix Construction, and Latent Dirichlet Allocation.

The authors identified computational constraints, data sparsity, risk of overfitting,
and ethical/privacy risks.

I am unclear with the authors mentioning "data-driven methods" for fine-tuning LDA parameters but do not describe them in detail

```python
# Download pdf
import requests

pdf_url = 'https://www.apa.org/pubs/journals/features/met-met0000104.pdf'
output_filename = 'MiningBigDataPaper_met-met0000104.pdf'

try:
    response = requests.get(pdf_url, stream=True)
    response.raise_for_status()

    with open(output_filename, 'wb') as pdf_file:
        for chunk in response.iter_content(chunk_size=8192):
            pdf_file.write(chunk)
    print(f"PDF downloaded successfully as '{output_filename}'")
except requests.exceptions.RequestException as e:
    print(f"Error downloading PDF: {e}")
```
