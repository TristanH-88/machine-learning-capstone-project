## Challenge 2: Create the sparse user-likes matrix
Create a **sparse matrix** that stores the data for the user-likes associations as described in the paper. The name of the user-likes matrix must be `M`, to keep the naming convention used in the paper. It is important that to store the data in a sparse matrix; you might need to research the topic of using sparse matrices in python (i.e. look at the `scipy.sparse` package and the `csr_matrix` class). *Hint: For this challenge,  you might find  the `merge` method from `pandas` library useful*.

Once you create the matrix print its shape on the answer summary cell.

```
#
# Your code goes here.
#

# Create mappings for user IDs and like IDs to numerical indices
user_to_idx = {user_id: idx for idx, user_id in enumerate(users['userid'].unique())}
like_to_idx = {like_id: idx for idx, like_id in enumerate(likes['likeid'].unique())}

# Keep track of original user/like IDs corresponding to matrix rows/columns
current_user_ids = users['userid'].unique()
current_like_ids = likes['likeid'].unique()

# Map the 'userid' and 'likeid' in the 'ul' dataframe to their numerical indices
ul['user_idx'] = ul['userid'].map(user_to_idx)
ul['like_idx'] = ul['likeid'].map(like_to_idx)

# Drop rows where mapping might have failed (shouldn't happen if all IDs are consistent)
ul_mapped = ul.dropna(subset=['user_idx', 'like_idx']).copy()

# Convert indices to integers
ul_mapped['user_idx'] = ul_mapped['user_idx'].astype(int)
ul_mapped['like_idx'] = ul_mapped['like_idx'].astype(int)

# Create the data, row indices, and column indices for the sparse matrix
data = np.ones(len(ul_mapped)) # A '1' indicates that the user liked the item
rows = ul_mapped['user_idx'].values
cols = ul_mapped['like_idx'].values

# Determine the dimensions of the sparse matrix
num_users = len(user_to_idx)
num_likes = len(like_to_idx)

# Create the sparse matrix M using csr_matrix
M = csr_matrix((data, (rows, cols)), shape=(num_users, num_likes))
```

```
#
# Answer summary cell:  Print the shape of the matrix M
#
print(f"Shape of the sparse user-likes matrix M: {M.shape}")
```

## Challege 3: Trim the sparse user-likes matrix M.
The paper calls for the matrix `M` to be trimmed (i.e. to remove users which did not like a minimum number of posts and posts that did not receive a minimum number of likes). In this challenge, you are expected to implement the trimming preprocessing step as described in the paper. The resulting matrix must be a sparse matrix. *Hint: you might want to research to index slice sparse matrices in python*

```
#
# Your code here.
#

min_user_likes = 5  # Minimum number of likes a user must have
min_like_users = 5  # Minimum number of users who must have liked an item

# Perform iterative trimming
while True:
    prev_num_users, prev_num_likes = M.shape

    # Trim users (rows) who don't have enough likes
    user_counts = M.sum(axis=1).A1
    users_to_keep_mask = user_counts >= min_user_likes
    M = M[users_to_keep_mask, :]
    current_user_ids = current_user_ids[users_to_keep_mask]

    # Trim likes (columns) that don't have enough users
    like_counts = M.sum(axis=0).A1
    likes_to_keep_mask = like_counts >= min_like_users
    M = M[:, likes_to_keep_mask]
    current_like_ids = current_like_ids[likes_to_keep_mask]

    current_num_users, current_num_likes = M.shape

    # If no rows or columns were removed in this iteration, stop
    if current_num_users == prev_num_users and current_num_likes == prev_num_likes:
        break

print(f"Trimming complete. Original shape: ({prev_num_users}, {prev_num_likes}), Final shape: {M.shape}")
```

```
#
# Answer summary cell:  Print the shape of the matrix M after trimming.
#
print(f"Shape of the sparse user-likes matrix M after trimming: {M.shape}")
```

## Challenge 4:  Split the sparse matrix into a training and testing matrices
To build a prediction model the paper calls for separating the data matrix (i.e. the user-like matrix) into a training and testing matrices. As part of this challenge, you need to create a training matrix `M_train` that includes the preferences of 80% of the users, and a testing matrix `M_test` that include the preferences of the remaining 20% of the users in the dataset. Both matrices `M_train` and `M_test` must be stored as sparse matrices. User will be randomly assigned to the training or testing set.

Print the shapes of the `M_train` and `M_test` matrices.




```
#
# Your code here.
#

num_users = M.shape[0]

# Generate a random permutation of user indices
user_indices = np.arange(num_users)
np.random.shuffle(user_indices)

# Determine the split point (80% for training, 20% for testing)
split_point = int(0.8 * num_users)

# Get training and testing user indices
train_user_indices = user_indices[:split_point]
test_user_indices = user_indices[split_point:]

# Create M_train and M_test by slicing the original sparse matrix M
M_train = M[train_user_indices, :]
M_test = M[test_user_indices, :]
```

```
#
# Answer summary cell: Print the shapes of M_train, M_test and M matrices.
#
print(f"Shape of the original trimmed matrix M: {M.shape}")
print(f"Shape of the training matrix M_train: {M_train.shape}")
print(f"Shape of the testing matrix M_test: {M_test.shape}")
```

## Challege 5 : Reduce the dimensionality of the M_train matrix using SVD

In this challenge, you must use singular value decomposition (SVD) to reduce the dimensionality of the training matrix `M_train`. Remember that `M_train` is encoded as a sparse matrix so you would need to use an appropriate implementation of the SVD for sparse matrices. Your output should be the matrices `U`, `S` and `Vt` that correspond to the left singular vectors, the singular values, and the right singular vectors, respectively. You must specify a parameter `k` that defines the number of reduced dimensions you want to keep.

Print the shapes of the `U`, `S` and `Vt` matrices and the value you selected for parameter `k`.



```
#
# Your code here.
#
k = 50  # Number of reduced dimensions

# Perform SVD on the sparse matrix M_train
# svds returns U, s (singular values as 1D array), Vt
U, s, Vt = svds(M_train, k=k)

# Convert singular values (s) into a diagonal matrix S
S = np.diag(s)
```

```
#
# Answer summary cell: Print the shapes of U, S and Vt; and the value you selected for parameter `k`.
#
print(f"Value selected for k: {k}")
print(f"Shape of U matrix: {U.shape}")
print(f"Shape of S matrix: {S.shape}")
print(f"Shape of Vt matrix: {Vt.shape}")
```

## Challege 6 : Use the varimax algorithm to rotate the SVD dimensions.
Use the varimax algorithm to rotate the SVD dimensions as described in the paper. A good python library that implements varimax algorithm is the `factor_analyzer` and its `Rotator` class (already installed in this notebook for your in the started cell above); i.e. using `rotator = Rotator(method='varimax')`. Store the varimax SVD into the matrices `U_rot` and `V_rot` respectively. *Hint: you can apply varimax of `Vt` to get `V_rot` and then multiply `M` by  `Vt_rot` to obtain the `U_rot` (as done in the paper); you might want to research hot to user the `Rotator` class*.

Print the shapes of the matrices `U_rot` and `V_rot`.


```
#
# Your code here.
#

# Add a very small amount of noise to Vt.T to prevent numerical instability
noise_level = 1e-10
Vt_noisy_T = Vt.T + np.random.normal(0, noise_level, Vt.T.shape)

# Initialize the Rotator with varimax method
rotator = Rotator(method='varimax')

# Rotate the noisy Vt.T to get V_rotated_loadings
# Vt has shape (k, num_likes). We need to transpose it to (num_likes, k) for the rotator.
V_rotated_loadings = rotator.fit_transform(Vt_noisy_T)

# Transpose back to get V_rot with shape (k, num_likes)
V_rot = V_rotated_loadings.T

# Calculate U_rot using M_train and the rotated loadings (V_rotated_loadings)
U_rot = M_train @ V_rotated_loadings
```

```
#
# Answer summary cell: Print the shapes of the matrices `U_rot` and `V_rot`
#
print(f"Shape of U_rot matrix: {U_rot.shape}")
print(f"Shape of V_rot matrix: {V_rot.shape}")
```

### Political Affiliation Target Variable Preparation

Before training the models, we need to extract the political affiliation for the users present in our training data (`U_rot`). We also need to handle any missing values in the 'political' column and ensure the target variable is correctly aligned with the feature matrix `U_rot`.

```
# Map back original user IDs to indices for U_rot alignment
# The user IDs corresponding to the *rows* of the trimmed M matrix are now in current_user_ids.

# Get the user IDs corresponding to the training set (U_rot rows)
train_final_user_ids = current_user_ids[train_user_indices]

# Extract political affiliation for these users, ensuring order aligns with U_rot
# Use .loc with the index to ensure correct alignment if users dataframe index is not ordered like train_final_user_ids
political_data_series = users.set_index('userid').loc[train_final_user_ids]['political']

# Handle missing values in political affiliation: remove users with NaN political data
non_missing_mask = ~political_data_series.isnull()

X_train_clean = U_rot[non_missing_mask]
y_train_clean = political_data_series[non_missing_mask].values.astype(int)

print(f"Shape of X_train_clean (features after filtering NaN political values): {X_train_clean.shape}")
print(f"Shape of y_train_clean (target after filtering NaN political values): {y_train_clean.shape}")
print(f"Number of political affiliation classes: {np.unique(y_train_clean).shape[0]}")
```

**FIRST MODEL TRAINING: Use the cells below to train the predictive model using the first model you selected.**

### Model 1: Logistic Regression

```
#
# Use this cell to train the first model you selected for challenge 7.
# Add additional cells as needed below this cell.
#
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score

# Initialize Logistic Regression model
log_reg = LogisticRegression(max_iter=1000, solver='liblinear') # liblinear is good for small datasets and L1/L2 regularization

# Define parameter grid for GridSearchCV
param_grid_lr = {
    'C': [0.001, 0.01, 0.1, 1, 10, 100],
    'penalty': ['l1', 'l2']
}

# Setup GridSearchCV
grid_search_lr = GridSearchCV(log_reg, param_grid_lr, cv=5, scoring='accuracy', n_jobs=-1, verbose=1)

# Train (fit) the model with cross-validation
grid_search_lr.fit(X_train_clean, y_train_clean)

# Get the best model
best_log_reg_model = grid_search_lr.best_estimator_

# Predict on the training data with the best model
y_train_pred_lr = best_log_reg_model.predict(X_train_clean)

# Calculate training accuracy
training_accuracy_lr = accuracy_score(y_train_clean, y_train_pred_lr)
```
